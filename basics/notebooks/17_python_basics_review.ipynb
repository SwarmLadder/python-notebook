## Examples

### 1. Choosing collections

Lists preserve order and duplicates, sets preserve uniqueness, tuples model fixed sequences, and dictionaries provide key-based lookup. Choose based on the operations needed for the problem.

```python
records = [{"name": "Ari", "score": 83},
           {"name": "Bea", "score": 97},
           {"name": "Cam", "score": 83}]
print([row["name"] for row in records])
print(sorted({row["score"] for row in records}))
print({row["name"]: row["score"] for row in records})
```

**Check your understanding:** Which structure would let you look up a score by student name quickly?

### 2. Returning results from functions

Design helpers that return useful data rather than printing everywhere. Format output at the edges of the program, after calculation. Test the helper independently.

```python
def average(values):
    return sum(values) / len(values) if values else None

def label(score):
    if score >= 90:
        return "Excellent"
    if score >= 70:
        return "Pass"
    return "Needs improvement"
print(average([80, 90]))
print(label(94))
```

**Check your understanding:** What should `label` do with an out-of-range score?

### 3. Handling malformed data

Real data may be missing, invalid, or outside accepted ranges. Validation should reject unsafe input or apply a documented skip policy. Explicit errors are better than incorrect results.

```python
def parse_score(text):
    try:
        score = float(text)
    except (ValueError, TypeError):
        return None
    if 0 <= score <= 100:
        return score
    return None
for raw in ["78", "-2", "oops", None]:
    print(repr(raw), "->", parse_score(raw))
```

**Check your understanding:** What additional rule might be needed for values like float('nan')?

### 4. Capstone integration

A well-structured mini application combines a collection of records, loops, functions, validation, and formatting. The final project additionally saves information to CSV.

```python
gradebook = {"Ari": [80, 90], "Bea": [95, 98], "Cam": [65, 75]}
def report(gradebook):
    rows = []
    for name, grades in gradebook.items():
        avg = sum(grades) / len(grades) if grades else None
        rows.append((name, avg))
    return sorted(rows)
for name, avg in report(gradebook):
    print(f"{name:<8} {avg:.1f}" if avg is not None else f"{name:<8} no grades")
```

**Check your understanding:** How would you save these records so the next session can reload them?
