## 1. Choosing the right collection

- List: ordered items you'll change or loop over. Tuple: a fixed record. Set: unique items and fast `in`. Dict: look values up by key.
- Choose by the question you'll ask most often: "what's at position 3?" (list), "is x in here?" (set), "what's the value for this key?" (dict).
- Many programs combine them: a list of dicts, or a dict of lists.

## 2. Functions and messy data

- Real data has blanks, extra spaces and bad values. Clean first (`strip`, `lower`), then convert.
- Skip or report the bad rows instead of crashing: `try`/`except ValueError` around the conversion.
- Return results from functions and keep printing outside them.

```python
rows = ["ada, 92", "ben, n/a", " cy ,77 "]
scores, bad = {}, []
for row in rows:
    name, score = (part.strip() for part in row.split(","))
    try:
        scores[name] = int(score)
    except ValueError:
        bad.append(name)
print(scores, bad)
```

## 3. Putting it together

- Break a task into small functions, each easy to test on its own.
- Chain them: parse → clean → compute → format.
- Test each piece with a tiny input whose answer you know.

```python
def parse(line):
    item, qty, price = line.split(":")
    return item, int(qty), float(price)

def line_total(item, qty, price):
    return qty * price

order = ["tea:2:3.50", "cake:1:4.00"]
print(sum(line_total(*parse(line)) for line in order))
```
