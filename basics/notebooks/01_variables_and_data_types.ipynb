## Concepts and guided examples

Each concept starts with an explanation and a runnable example. The examples use Python's standard library so they work in a normal Jupyter Python 3 environment.

### 1. Names and numeric types

Valid names can include letters, underscores, and digits (not as the first character). Use `snake_case` for readability. Integers represent whole numbers, while floats approximate real numbers using binary floating point.

```python
user_age = 26
price = 19.95
print(type(user_age).__name__)
print(type(price).__name__)
print(0.1 + 0.2)
```

**Check your understanding:** Why should you avoid comparing money calculations using exact float equality?

### 2. Text, booleans, and missing values

Strings represent Unicode text and have quotation marks. Booleans are `True` and `False` without quotes. `None` indicates an intentional absence of a value, often used as a sentinel.

```python
name = "Ada"
is_registered = True
phone_number = None
for value in [name, is_registered, phone_number]:
    print(repr(value), type(value).__name__)
print(phone_number is None)
```

**Check your understanding:** How is `"False"` different from `False`?

### 3. Conversions and invalid input

The built-ins `int`, `float`, and `str` make new values of the requested type when conversion is valid. Converting an invalid numeric string raises `ValueError`; do not assume arbitrary input is clean.

```python
quantity = int("12")
unit_price = float("7.50")
print(f"Total: {quantity * unit_price:.2f}")
print(str(quantity) + " items")
```

**Check your understanding:** What exception would `int("twelve")` raise?

### 4. Value, identity, and mutability

`==` compares values, while `is` tests whether two references point to the very same object. Use `is None`. Lists are mutable and can be changed in place; strings and numbers are immutable.

```python
a = [1, 2]
b = a
c = [1, 2]
print(a == c, a is c)
b.append(3)
print(a)
print(b is a)
```

**Check your understanding:** Why did `a` change when only `b` was modified?
