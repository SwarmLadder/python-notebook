## 2. Whole numbers: int

- An `int` is a whole number of any size. `1_000_000` is the same as `1000000`.

- `//` is floor division (`7 // 2` is `3`), `%` is the remainder (`7 % 2` is `1`), and `**` is power (`2 ** 10` is `1024`).

```python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
print(type(7 // 2).__name__, type(7 / 2).__name__)
```

## 3. Decimals: float

- A `float` has a decimal point or exponent: `3.0`, `0.5`, `1.5e3` (which is `1500.0`).
- Mixing an int and a float gives a float: `3 + 0.5` is `3.5`.
- Floats are close approximations: `0.1 + 0.2` is `0.30000000000000004`. Use `round(x, 2)` for display and `math.isclose(a, b)` to compare, never `==`.

```python
print(0.1 + 0.2)
print(round(0.1 + 0.2, 2))
print(0.1 + 0.2 == 0.3, math.isclose(0.1 + 0.2, 0.3))
```

## 4. Text: str

- A `str` is text in single or double quotes: `'hi'` and `"hi"` are the same.
- `+` joins strings, `*` repeats them, and `len(s)` counts the characters. `s[0]` is the first character and `s[-1]` the last.
- An f-string puts values into text: `f"{name} is {age}"`. Add `:.2f` to show two decimals: `f"{7.5:.2f}"` is `"7.50"`.

```python
first, last = "Ada", "Lovelace"
full = first + " " + last
print(full, len(full), full[0], full[-1])
print("ab" * 3)
print(f"{first} owes ${7.5:.2f}")
```

## 5. True, False and None

- `True` and `False` are capitalised and have no quotes. Comparisons such as `==`, `!=`, `<` and `>=` produce them.
- In an `if`, `0`, `0.0`, `""`, `[]` and `None` count as false; almost everything else counts as true. So `bool("False")` is `True`: it's a non-empty string.
- `None` means "no value". Test for it with `is None`, not `== None`.

```python
print(3 > 2, 3 == 2)
print(bool(0), bool(""), bool("False"), bool([0]))
nickname = None
print(nickname is None)
```

## 6. Converting between types

- `int()`, `float()`, `str()` and `bool()` make a new value of that type: `int("12")` is `12`, `str(5) + "!"` is `"5!"`.
- `int(3.9)` cuts off the decimals and gives `3`; it does not round. `round()` rounds halves to the nearest even number: `round(2.5)` is `2`.
- Text that isn't a number raises `ValueError`: `int("twelve")`, and also `int("4.2")`. Spaces around a number are fine: `int(" 7 ")` is `7`.

```python
print(int("12") + 1, float("7.50") * 2, int(3.9), str(5) + "!")
print(type(3.0).__name__)
try:
    int("twelve")
except ValueError as error:
    print("ValueError:", error)
```

## 7. Same value or same object

- `==` asks "equal values?". `is` asks "the very same object?". Use `is` only with `None`.
- `b = a` does not copy: both names point to one object.
- Lists can change in place (they're mutable), so a change made through `b` is visible through `a`. Numbers and strings never change in place (they're immutable).

```python
a = [1, 2]
b = a
c = [1, 2]
print(a == c, a is c, a is b)
b.append(3)
print(a)
```
