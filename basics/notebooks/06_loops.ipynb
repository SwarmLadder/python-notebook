## Essential truth

A for loop takes one item at a time. A while loop tests its condition before each pass. Either body can run zero times. An accumulator must keep its value between passes.

## 1. for over a collection

- `for item in items:` runs the block once per item, in order. It works on lists, strings (one character at a time), tuples, sets and dicts (keys).
- The loop variable keeps the last value after the loop ends.
- Don't change a list while looping over it; build a new list instead.

```python
for fruit in ["apple", "pear", "mango"]:
    print(fruit.upper())
for letter in "hey":
    print(letter, end=" ")
```

## 2. range

- `range(stop)` counts `0, 1, …, stop-1`. `range(start, stop)` starts at `start`.
- `range(start, stop, step)` jumps by `step`; a negative step counts down: `range(5, 0, -1)` is `5, 4, 3, 2, 1`.
- `stop` is never included.

```python
print(list(range(5)))
print(list(range(2, 6)))
print(list(range(0, 20, 5)))
print(list(range(5, 0, -1)))
```

## 3. Accumulating results

- Start a total before the loop (`total = 0`, `result = []`), update it inside, use it after.
- Count with `count += 1`; build with `result.append(x)`; multiply with `product *= x` starting from `1`.
- Track a best-so-far value: start with the first item, then replace it when you find a better one.

```python
total = 0
product = 1
for n in range(1, 6):
    total += n
    product *= n
print(total, product)
```

## 4. while loops

- `while condition:` repeats as long as the condition is true. Use it when you don't know the number of steps in advance.
- Something inside the loop must change the condition, or it never stops.
- Pressing Stop in the toolbar ends a loop that runs forever.

```python
n = 27
steps = 0
while n != 1:
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print(steps)
```

## 5. break, continue and enumerate

- `break` leaves the loop at once; `continue` skips to the next item.
- `for i, item in enumerate(items):` gives each position and item; `enumerate(items, start=1)` counts from 1.
- Nested loops run the inner loop fully for every step of the outer one.

```python
for n in [4, 9, -2, 7]:
    if n < 0:
        print("first negative:", n)
        break
for i, name in enumerate(["Ada", "Linus"], start=1):
    print(i, name)
for row in range(1, 4):
    print("*" * row)
```
