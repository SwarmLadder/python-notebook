## Concepts and guided examples

Each concept starts with an explanation and a runnable example. The examples use Python's standard library so they work in a normal Jupyter Python 3 environment.

### 1. for loops and range

`for` visits each item of an iterable. `range(start, stop, step)` produces a sequence of integers excluding the stop value. Negative steps count down.

```python
for fruit in ["apple", "pear", "mango"]:
    print(fruit)
for n in range(1, 6):
    print(n, n*n)
print(list(range(10, 0, -2)))
```

**Check your understanding:** What expression generates odd numbers from 1 through 19?

### 2. Accumulators

A running total, product, or count starts with an initial state and changes during each iteration. Explain what the accumulator represents after each step.

```python
prices = [9.99, 5.50, 4.00]
total = 0.0
for price in prices:
    total += price
    print(f"Running total: {total:.2f}")
print(f"Final: {total:.2f}")
```

**Check your understanding:** What should the total be when the list is empty?

### 3. while and loop control

`while` repeats until its condition becomes false. Without state changes, a loop can run forever. `continue` skips to the next iteration; `break` exits the nearest loop.

```python
count = 3
while count > 0:
    print(count)
    count -= 1
print("Go!")
for n in range(1, 8):
    if n == 4:
        continue
    if n == 7:
        break
    print(n)
```

**Check your understanding:** What if `count -= 1` were removed?

### 4. Nested loops and enumerate

A nested loop runs its inner body once for each combination of outer and inner values. `enumerate` provides positions and items without manually maintaining an index.

```python
for row in range(1, 4):
    for col in range(1, 4):
        print(row * col, end=" ")
    print()
for i, item in enumerate(["red", "blue"], start=1):
    print(i, item)
```

**Check your understanding:** How many products does the nested grid calculate?
