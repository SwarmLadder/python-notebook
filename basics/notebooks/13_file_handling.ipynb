### 1. Paths and the working directory

Jupyter may start in a different working directory depending on how it is launched. Relative paths resolve from the kernel's `Path.cwd()`. `pathlib` combines paths without hardcoded platform separators.

```python
from pathlib import Path
print(Path.cwd())
print(Path("basics") / "data" / "students.csv")
print(Path("records.csv").suffix)
```

**Check your understanding:** How would changing the working directory affect a relative file path?

### 2. Writing text safely

Mode `w` overwrites existing content; `a` appends; `r` reads. `with` ensures file resources are released. Using a temporary directory in teaching examples avoids accidentally changing course files.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
with TemporaryDirectory() as folder:
    path = Path(folder) / "notes.txt"
    path.write_text("first\nsecond\n", encoding="utf-8")
    print(path.read_text(encoding="utf-8"))
    with path.open("a", encoding="utf-8") as handle:
        handle.write("third\n")
    print(path.read_text(encoding="utf-8"))
```

**Check your understanding:** Why is `w` dangerous for a file you want to preserve?

### 3. Processing lines

Line iteration makes it possible to process large files incrementally. `strip` removes leading and trailing whitespace; use `rstrip` if left indentation matters. Always decide how blank lines should be treated.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
with TemporaryDirectory() as folder:
    path = Path(folder) / "items.txt"
    path.write_text(" apple \n\n banana \n", encoding="utf-8")
    with path.open(encoding="utf-8") as handle:
        items = [line.strip() for line in handle if line.strip()]
    print(items)
```

**Check your understanding:** What does the `if line.strip()` part filter out?

### 4. Writing and reading CSV

CSV files store text fields separated and quoted according to CSV rules. `csv.DictWriter` and `csv.DictReader` handle formatting and columns. Numeric fields loaded from CSV must be converted before calculations.

```python
import csv
from pathlib import Path
from tempfile import TemporaryDirectory
rows = [{"name": "Ari", "grade": 88}, {"name": "Bea", "grade": 92}]
with TemporaryDirectory() as folder:
    path = Path(folder) / "students.csv"
    with path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["name", "grade"])
        writer.writeheader()
        writer.writerows(rows)
    with path.open("r", newline="", encoding="utf-8") as f:
        loaded = list(csv.DictReader(f))
    print(loaded)
    print(sum(int(row["grade"]) for row in loaded))
```

**Check your understanding:** Why is newline='' recommended when opening a CSV file?
