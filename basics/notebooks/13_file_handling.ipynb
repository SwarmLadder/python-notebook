## 1. Writing and reading text

- `with open(path, "w", encoding="utf-8") as f:` opens a file for writing and closes it when the block ends. `"w"` replaces the file; `"a"` appends.
- `f.write(text)` writes exactly what you give it: add `"\n"` yourself to end a line.
- `Path(path).read_text(encoding="utf-8")` reads the whole file; `.splitlines()` splits it into lines without the `\n`.

```python
folder = Path(tempfile.mkdtemp())   # a scratch folder for this lesson
path = folder / "notes.txt"
with open(path, "w", encoding="utf-8") as f:
    f.write("first\n")
    f.write("second\n")
print(path.read_text(encoding="utf-8").splitlines())
```

## 3. CSV files

- `csv.DictWriter(f, fieldnames=[...])` writes rows from dicts; call `writeheader()` first.
- `csv.DictReader(f)` reads each row as a dict whose values are all strings: convert numbers yourself.
- Open CSV files with `newline=""` so line endings come out right on every system.

```python
path = Path(tempfile.mkdtemp()) / "grades.csv"
with open(path, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "grade"])
    writer.writeheader()
    writer.writerows([{"name": "Ada", "grade": 92}, {"name": "Ben", "grade": 78}])
print(path.read_text(encoding="utf-8"))
with open(path, newline="", encoding="utf-8") as f:
    print(list(csv.DictReader(f)))
```
