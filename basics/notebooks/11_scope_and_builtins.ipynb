## 1. Local and global names

- A name assigned inside a function is local: it exists only during that call.
- A function can read a global name, but assigning to it creates a new local name instead. Pass values in and return results rather than changing globals.
- Each call starts fresh: local names don't remember earlier calls.

```python
rate = 0.2
def with_tax(price):
    total = price * (1 + rate)   # reads the global rate
    return total
print(with_tax(10))
try:
    print(total)
except NameError as error:
    print("NameError:", error)
```

## 2. Built-ins you'll use every day

- `len`, `sum`, `min`, `max`, `abs`, `round`, `sorted` work on most sequences of numbers or text.
- `min` and `max` take `key=`: `max(words, key=len)` finds the longest word.
- `any(...)` is true if at least one item is true; `all(...)` if every item is.

```python
words = ["fig", "banana", "kiwi"]
print(len(words), max(words, key=len), sorted(words, key=len))
print(abs(-3), round(2.675, 2), any(w.startswith("b") for w in words), all(len(w) > 2 for w in words))
```

## 3. zip and enumerate

- `zip(a, b)` walks two sequences together, giving pairs; it stops at the shorter one.
- `enumerate(items, start=1)` gives position and item.
- Combine them: `for i, (name, score) in enumerate(zip(names, scores), start=1):`.

```python
names = ["Ada", "Linus"]
scores = [92, 78]
for i, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"{i}. {name}: {score}")
```
