## Examples

### 1. Local and global variables

Names assigned inside a function are local by default. Functions can read globals, but passing values and returning results usually produces safer, reusable code. Avoid depending on mutable global state.

```python
tax_rate = 0.08
def total_with_tax(amount):
    result = amount * (1 + tax_rate)
    return result
print(total_with_tax(100))
```

**Check your understanding:** How would you rewrite this to accept tax_rate as a parameter?

### 2. Name lookup and shadowing

Python checks local, enclosing, global, then built-in scopes (LEGB). A variable called `sum` would hide the `sum()` function in that scope. Choose names that do not collide with built-ins.

```python
label = "global"
def outer():
    label = "enclosing"
    def inner():
        return label
    return inner()
print(outer())
print(label)
```

**Check your understanding:** What would the inner function return if outer did not assign label?

### 3. Numerical built-ins

Useful built-ins can express a full transformation in a line. `min` and `max` fail on empty inputs unless a suitable default is supplied. `round` uses Python's numeric rounding semantics.

```python
values = [-5, 2, 9, 4]
print(len(values), sum(values))
print(min(values), max(values))
print(abs(values[0]), sorted(values))
print(round(2.345, 2))
```

**Check your understanding:** How could you calculate an average while handling an empty list?

### 4. Iteration helpers

`enumerate` supplies index and item; `zip` pairs corresponding elements and stops at the shortest iterable. `any` and `all` inspect whether at least one or every element meets a condition.

```python
names = ["Ari", "Bea", "Cam"]
marks = [82, 91, 77]
for position, (name, score) in enumerate(zip(names, marks), start=1):
    print(position, name, score)
print(any(score >= 90 for score in marks))
print(all(score >= 50 for score in marks))
```

**Check your understanding:** What happens to extra elements if the lists passed to zip differ in length?
