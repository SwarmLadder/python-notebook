## Examples

### 1. Code cells and output

A Jupyter kernel runs Python statements and remembers assigned variables between executions. `print` explicitly displays a value, while the last expression in a cell is often displayed automatically. The order of cell execution matters.

```python
print("Hello, Python!")
print("2 + 3 =", 2 + 3)
print("Welcome to the course")
```

**Check your understanding:** Which lines would still work if this were a regular `.py` script?

### 2. Names and execution order

A name begins to refer to a value when an assignment executes. The same name can later refer to a new value. Restarting a kernel clears all of these assignments, which is why reproducible notebooks must run successfully from top to bottom.

```python
language = "Python"
print(language)
language = "Python 3"
print(language)
```

**Check your understanding:** What happens if you restart the kernel before accessing `language` in a different cell?

### 3. Comments and basic expressions

Lines beginning with `#` are explanatory comments. Comments should explain intent rather than repeat what the next line says. Operators are evaluated according to precedence; parentheses make your meaning easier to see.

```python
# Compute the price of three tickets.
ticket_price = 12
tickets = 3
print("Total:", ticket_price * tickets)
print((2 + 3) * 4)
```

**Check your understanding:** Why can `(2 + 3) * 4` differ from `2 + 3 * 4`?

### 4. Indentation and syntax

Python relies on indentation to group statements beneath `if`, loops, and function definitions. A colon introduces a suite of statements and four spaces is the standard indentation style. Python is also case-sensitive.

```python
temperature = 23
if temperature >= 20:
    print("A warm day")
else:
    print("A cool day")
print("Done")
```

**Check your understanding:** Which print calls execute when `temperature` is 10?
