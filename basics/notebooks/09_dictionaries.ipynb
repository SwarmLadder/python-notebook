# 09 — Dictionaries

Dictionaries are mappings from unique hashable keys to values. They let you model small structured records and build efficient lookup tables.

*Basics · 60–90 min*

## You will learn

- Create, retrieve, and update mappings
- Use get for optional keys
- Iterate through items and nested records
- Count frequencies

## Examples

### 1. Keys and lookups

A dictionary is written with `{key: value}` entries. Accessing a missing key using brackets raises `KeyError`; `get` can supply a default. Reassigning an existing key replaces its associated value.

```python
student = {"name": "Aisha", "age": 20, "grade": 91}
print(student["name"])
print(student.get("email", "Not provided"))
student["grade"] = 94
print(student)
```

**Check your understanding:** What is the difference between `student['email']` and `student.get('email')`?

### 2. Updating and iteration

The `items()` method produces key–value pairs; `keys()` and `values()` produce views. `update` adds or replaces entries and `pop` removes a key and returns its value.

```python
inventory = {"pen": 10, "book": 5}
inventory["pencil"] = 8
inventory.update({"book": 7})
for item, quantity in inventory.items():
    print(item, quantity)
print(inventory.pop("pen"))
```

**Check your understanding:** How could you avoid an error when popping an absent key?

### 3. Nested records

A dictionary can contain other dictionaries and lists, allowing student records or product catalogs. Use consistent inner schemas and validate numeric values before averaging.

```python
students = {"S01": {"name": "Mira", "scores": [86, 90]},
            "S02": {"name": "Noah", "scores": [71, 79]}}
for student_id, row in students.items():
    print(student_id, row["name"], sum(row["scores"]) / len(row["scores"]))
```

**Check your understanding:** How would you access Mira's second score?

### 4. Counting frequencies

Counting uses a dictionary accumulator: get the previous count (defaulting to zero), add one, and save it back. This pattern generalizes to words, categories, and grouped metrics.

```python
counts = {}
for word in "to be or not to be".split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
```

**Check your understanding:** What should a frequency dictionary contain if the input is empty?
