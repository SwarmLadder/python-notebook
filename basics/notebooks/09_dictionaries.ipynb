## 3. Counting and grouping

- Count with `counts[key] = counts.get(key, 0) + 1`.
- `collections.Counter(items)` counts in one step, and `.most_common(n)` gives the top `n`.
- Group values in lists: `groups.setdefault(key, []).append(value)`.

```python
words = "the cat and the hat".split()
counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1
print(counts)
print(Counter(words).most_common(1))
groups = {}
for w in words:
    groups.setdefault(len(w), []).append(w)
print(groups)
```

## 4. Combining and inverting

- `dict(zip(keys, values))` pairs two lists into a dict.
- `a | b` makes a merged copy where `b` wins on shared keys; `a.update(b)` changes `a` in place.
- Invert a dict with `{v: k for k, v in d.items()}`. This only works when the values are unique.

```python
names = ["Ada", "Linus"]
ages = [36, 28]
print(dict(zip(names, ages)))
print({"a": 1, "b": 2} | {"b": 9})
print({v: k for k, v in {"a": 1, "b": 2}.items()})
```
