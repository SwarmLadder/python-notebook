## 1. Reading errors

- An exception stops the program and prints a traceback. Read its last line first: the error type and message.
- Common types: `ValueError` (right type, bad value), `TypeError` (wrong type), `KeyError` (missing dict key), `IndexError` (position out of range), `ZeroDivisionError`, `FileNotFoundError`.
- The traceback's lowest line in your code is where it happened.

```python
for code in ["int('x')", "1 / 0", "[1][5]", "{}['k']", "'a' + 1"]:
    try:
        eval(code)
    except Exception as error:
        print(f"{code:10} -> {type(error).__name__}")
```

## 2. try and except

- Put the risky line in `try:`; handle a specific failure in `except ValueError:`. Never use a bare `except:`.
- Keep the `try` block small, so you only catch the error you expect.
- `else:` runs when no error happened; `finally:` always runs (for cleanup).

```python
def parse_number(text):
    try:
        return float(text)
    except ValueError:
        return None

print(parse_number("4.5"), parse_number("four"))
```

## 3. Raising your own errors

- `raise ValueError("clear message")` stops a function when its input makes no sense.
- Say what was wrong and what's allowed: `"percentage must be 0–100, got 120"`.
- Check inputs at the top of the function (a guard), then do the work.

```python
def check_percentage(value):
    if not 0 <= value <= 100:
        raise ValueError(f"percentage must be 0-100, got {value}")
    return value

print(check_percentage(40))
try:
    check_percentage(120)
except ValueError as error:
    print(error)
```
