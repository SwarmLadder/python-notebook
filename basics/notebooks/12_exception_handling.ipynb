## Examples

### 1. Common exceptions

Syntax errors prevent parsing, whereas runtime exceptions arise during execution. Examples include `ValueError`, `TypeError`, `ZeroDivisionError`, `IndexError`, and `KeyError`. Reproduce errors with a small example before fixing them.

```python
for text in ["12", "oops", "-3"]:
    try:
        print(int(text))
    except ValueError as error:
        print("Invalid number:", error)
```

**Check your understanding:** What error occurs when you divide by zero?

### 2. Narrow try/except blocks

Catch errors around the smallest operation likely to fail. Broad exception handlers may accidentally conceal a bug. Keep successful values separate from malformed values when importing data.

```python
valid = []
for text in ["10", "oops", "25"]:
    try:
        valid.append(int(text))
    except ValueError:
        print("Skipping:", text)
print(valid)
```

**Check your understanding:** Would you skip or reject the entire data file? Discuss the trade-off.

### 3. else and finally

An `else` on a try statement runs only when no exception occurs; `finally` runs even when returning or raising. A `with` context manager is generally preferable for handling file cleanup.

```python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        return None
    else:
        return result
    finally:
        print("Attempt finished")
print(divide(12, 3))
print(divide(12, 0))
```

**Check your understanding:** Which lines print when denominator is zero?

### 4. Raising your own errors

Validate assumptions near a function's entry point. A clear `ValueError` can explain an impossible range or invalid choice; callers then decide how to recover.

```python
def percent_of(amount, percentage):
    if not 0 <= percentage <= 100:
        raise ValueError("percentage must be between 0 and 100")
    return amount * percentage / 100
print(percent_of(200, 15))
try:
    percent_of(200, 150)
except ValueError as exc:
    print(exc)
```

**Check your understanding:** Why is a descriptive error better than silently returning a wrong answer?
