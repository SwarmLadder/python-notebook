## Examples

### 1. Tuples and unpacking

A tuple is an ordered, immutable sequence. A single-element tuple needs a comma: `(42,)`. Unpacking separates values into variables, and swapping can happen without a temporary variable.

```python
position = (44.43, 26.10)
latitude, longitude = position
print(latitude, longitude)
print(type((42,)).__name__)
x, y = 1, 2
x, y = y, x
print(x, y)
```

**Check your understanding:** Can a tuple contain a mutable list? What would immutability mean then?

### 2. Sets and uniqueness

Sets contain unique hashable values and are optimized for membership checks on average. Their iteration order is not a substitute for a sorted list. Use `set()` to make an empty set; `{}` makes an empty dictionary.

```python
numbers = [2, 2, 3, 5, 5, 7]
unique = set(numbers)
unique.add(11)
unique.discard(100)
print(sorted(unique))
print(3 in unique)
```

**Check your understanding:** When could removing duplicates be undesirable?

### 3. Set operations

`|` means union, `&` intersection, `-` difference, and `^` symmetric difference. These operators are useful for comparing participants, permissions, or tags across groups.

```python
python = {"Ada", "Ben", "Cleo"}
sql = {"Ben", "Dina", "Cleo"}
print(sorted(python | sql))
print(sorted(python & sql))
print(sorted(python - sql))
print(sorted(python ^ sql))
```

**Check your understanding:** Which expression finds people who study Python but not SQL?

### 4. Choosing the right collection

Use a tuple for fixed coordinates, a list for a changing order of tasks, and a set for a unique membership collection. Converting a set back to a list does not recover original ordering; sort if consistent order matters.

```python
coordinate = (10, 20)
todo = ["learn", "practice"]
tags = {"beginner", "python", "beginner"}
print(coordinate)
print(todo)
print(sorted(tags))
```

**Check your understanding:** Which collection best represents a sequence where duplicates and order both matter?
