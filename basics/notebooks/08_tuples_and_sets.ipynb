## 1. Tuples

- A tuple is a fixed sequence in `( )`: `point = (3, 4)`. It can't be changed after it's made.
- Unpack it into names: `x, y = point`. A function can return several values as a tuple.
- A one-item tuple needs a comma: `(5,)`. Without it, `(5)` is just the number 5.

```python
point = (3, 4)
x, y = point
print(x, y, len(point))
def min_max(values):
    return min(values), max(values)
low, high = min_max([4, 9, 1])
print(low, high, type((5,)).__name__, type((5)).__name__)
```

## 2. Sets

- A set holds unique items in `{ }`, with no order: `{1, 2, 2}` is `{1, 2}`. An empty set is `set()`, because `{}` is an empty dict.
- `set(items)` removes duplicates; `x in s` is fast.
- `s.add(x)` adds an item; `s.discard(x)` removes it if present.

```python
tags = {"python", "code", "python"}
print(tags, len(tags))
print(set([3, 1, 3, 2]))
tags.add("learn")
tags.discard("code")
print(sorted(tags))
```

## 3. Set operations

- `a & b` items in both; `a | b` items in either; `a - b` items in `a` only; `a ^ b` items in exactly one.
- `a <= b` asks whether every item of `a` is in `b` (a subset).
- Sort a set to show it in a stable order: `sorted(a & b)`.

```python
python = {"Ada", "Linus", "Grace"}
sql = {"Grace", "Edgar"}
print(sorted(python & sql), sorted(python | sql), sorted(python - sql), sorted(python ^ sql))
print({"Grace"} <= sql)
```
