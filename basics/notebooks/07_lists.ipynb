## 2. Changing a list

```python
cart = ["tea", "milk"]
cart.append("bread")
cart.insert(0, "eggs")
cart.remove("milk")
last = cart.pop()
print(cart, last)
```

## 3. Sorting and building

- `sorted(items)` returns a new sorted list; `items.sort()` sorts in place. Add `reverse=True` for largest first.
- `sum`, `min`, `max` work on lists of numbers.

A **list comprehension** collects one result per matching item. To read `[s + 1 for s in scores if s >= 70]`, start at `for`: visit each score in order, keep it if it meets the condition, then append the value of `s + 1` to a fresh list. Removing the filter keeps every item. No match produces an empty list. Compare both implementations below.

```python
scores = [72, 90, 60, 85]
print(sorted(scores), sorted(scores, reverse=True), scores)
print(sum(scores), min(scores), max(scores))
adjusted = []
for s in scores:
    if s >= 70:
        adjusted.append(s + 1)
print(adjusted)
print([s + 1 for s in scores if s >= 70])
```

## 4. Copies and nested lists

- `b = a` shares one list. `b = a.copy()` (or `a[:]`) makes a separate copy.
- A list can hold lists: `grid[1][0]` is row 1, column 0.
- Visit each row with an outer loop and each item with an inner loop. Empty rows contribute no items.

```python
a = [1, 2]
b = a.copy()
b.append(3)
print(a, b)
grid = [[1, 2], [3, 4]]
print(grid[1][0])
for row in grid:
    for item in row:
        print(item)
```
