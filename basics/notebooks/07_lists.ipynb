## Examples

### 1. Access and update items

List indexes begin at zero and support negative indexing. Slices create new lists; direct element assignments update the original. A list can store different types, although consistent element types simplify applications.

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[0], fruits[-1])
print(fruits[:2])
fruits[1] = "pear"
print(fruits)
```

**Check your understanding:** What happens when you access an index outside the list?

### 2. Appending, extending, and removing

`append` adds one object; `extend` adds every element from an iterable. `insert` chooses a position, `pop` removes and returns an item, and `remove` deletes the first matching value.

```python
tasks = ["review", "practice"]
tasks.append("quiz")
tasks.extend(["commit", "reflect"])
tasks.insert(0, "open notebook")
completed = tasks.pop(1)
print("Completed:", completed)
print(tasks)
```

**Check your understanding:** What is the difference between append([1,2]) and extend([1,2])?

### 3. Ordering and searching

`sorted` returns a new list while `list.sort()` changes the existing list and returns `None`. `in`, `count`, and `index` answer membership and location questions. Avoid changing a list's length while iterating over it.

```python
scores = [72, 95, 88, 72]
print(sorted(scores))
print("Original:", scores)
scores.sort(reverse=True)
print("Descending:", scores)
print(scores.count(72))
```

**Check your understanding:** Why is `scores = scores.sort()` incorrect?

### 4. Aliases and nested lists

Assigning one list to another variable makes a second reference to the same list. `copy` creates a shallow copy whose nested mutable objects are still shared. A grid is a list of lists.

```python
original = [[1, 2], [3, 4]]
shallow = original.copy()
shallow[0].append(99)
print(original)
print(shallow)
print(original[1][0])
```

**Check your understanding:** Which objects are shared between the two outer lists?
