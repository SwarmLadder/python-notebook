### 1. Specifications and counters

Define the question precisely. Counting positives requires deciding whether zero is positive and how to treat an empty list. Then maintain a counter whose meaning is clear after each iteration.

```python
def count_positive(values):
    count = 0
    for value in values:
        if value > 0:
            count += 1
    return count
assert count_positive([-3, 0, 2, 8]) == 2
print(count_positive([1, -1, 4]))
```

### 2. Finding an extremum

When finding a maximum manually, initialize from the first item rather than zero; otherwise a negative-only input produces a wrong answer. Define an explicit policy for empty inputs.

```python
def manual_max(values):
    if not values:
        raise ValueError("expected at least one value")
    best = values[0]
    for value in values[1:]:
        if value > best:
            best = value
    return best
assert manual_max([-8, -2, -5]) == -2
print(manual_max([4, 1, 9]))
```

**Check your understanding:** Why is starting `best=0` unsafe?

### 3. Frequency tables

For each character or word, use a dictionary to track how often it appeared. Decide in advance whether text should be case-insensitive and whether spaces/punctuation count.

```python
def frequencies(text):
    counts = {}
    for character in text.casefold():
        if character.isalpha():
            counts[character] = counts.get(character, 0) + 1
    return counts
print(frequencies("Hello, HELLO!"))
```

**Check your understanding:** Would hyphens be counted by this function?

### 4. Prime checking

A prime is a whole number above 1 divisible only by 1 and itself. Testing potential divisors up to the square root is sufficient, since any larger factor must pair with a smaller one.

```python
def is_prime(number):
    if number < 2:
        return False
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True
assert is_prime(29)
assert not is_prime(1)
print([n for n in range(2, 20) if is_prime(n)])
```

**Check your understanding:** Why is checking divisors only up to sqrt(n) valid?
