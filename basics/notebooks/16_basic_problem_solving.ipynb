## 1. From problem to steps

- Write down the input, the output and two or three examples with known answers before coding.
- Most simple algorithms are one loop with something tracked: a count, a total, a best-so-far or a result list.
- Handle the edge cases on purpose: an empty list, one item, ties, negative numbers.

```python
def count_long(words, n):
    """How many words have more than n letters? count_long(["hi", "hello"], 3) -> 1"""
    count = 0
    for w in words:
        if len(w) > n:
            count += 1
    return count
print(count_long(["hi", "hello", "hey"], 2), count_long([], 2))
```

## 2. Finding the best

- Start the best-so-far with the first item (not 0: the values might all be negative).
- Compare each item with the best and replace it when the new one wins. With `>`, the first of a tie wins; with `>=`, the last.
- Track the position too if you need to know where it was.

```python
temps = [-5, -2, -9]
warmest = temps[0]
for t in temps[1:]:
    if t > warmest:
        warmest = t
print(warmest)
```

## 3. Frequency and search

- Count things with a dict: `counts[x] = counts.get(x, 0) + 1`.
- To find duplicates, remember what you've seen in a set.
- Check a number for primality by trying divisors up to its square root.

```python
seen, duplicates = set(), []
for x in [3, 1, 3, 2, 1]:
    if x in seen and x not in duplicates:
        duplicates.append(x)
    seen.add(x)
print(duplicates)
```
