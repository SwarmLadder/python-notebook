## Examples

### 1. Arithmetic operators

Use `+`, `-`, `*`, `/`, `//`, `%`, and `**` for numeric work. `/` produces floating-point division, `//` is floor division, and `%` returns a remainder consistent with the divisor. Watch negative values.

```python
a, b = 17, 5
print(a + b, a - b, a * b)
print(a / b, a // b, a % b)
print(2 ** 5)
print(-7 // 3)
```

**Check your understanding:** What is the connection between quotient, remainder, and the dividend?

### 2. Comparisons and chains

`==`, `!=`, `<`, `<=`, `>`, and `>=` return booleans. Python supports readable chained comparisons such as `0 <= score <= 100`. They are ideal for checking a bounded range.

```python
age = 19
score = 83
print(age >= 18)
print(0 <= score <= 100)
print(score != 100)
```

**Check your understanding:** What changes for score -1 and score 101?

### 3. Boolean expressions

`and`, `or`, and `not` operate using truthiness and may return operands rather than literal booleans. Python short-circuits: it evaluates the right side only if necessary.

```python
has_ticket = True
has_id = False
print(has_ticket and has_id)
print(has_ticket or has_id)
username = ""
print(username or "Guest")
```

**Check your understanding:** Why does `False and (1/0)` not raise an error?

### 4. Membership and precedence

Use `in` for collection membership and text substrings. Augmented assignment such as `count += 1` is common in loops. Parentheses document the intended order of operations.

```python
count = 2
count += 3
print(count)
print("py" in "python")
print(3 in [1, 2, 3])
print(2 + 3 * 4, (2 + 3) * 4)
```

**Check your understanding:** Should `^` or `**` be used for exponentiation in Python?
