## 1. Arithmetic

- `+ - *` add, subtract and multiply. `/` divides and always gives a float.
- `//` divides and rounds down to a whole number; `%` gives the remainder. For any `a` and `b`: `a == (a // b) * b + a % b`.
- `**` is power (`2 ** 3` is `8`). Python follows maths order: `**`, then `* / // %`, then `+ -`. Use parentheses to make the order obvious.

```python
print(17 / 5, 17 // 5, 17 % 5, 2 ** 3)
print(2 + 3 * 4, (2 + 3) * 4)
print(-7 // 2, -7 % 2)   # rounds down, so -4 and 1
```

```python
score = 87
print(score >= 50, score == 100, 0 <= score <= 100)
print("apple" < "banana", "Zebra" < "apple")
```

## 3. and, or, not

- `a and b` is true only when both are; `a or b` when at least one is; `not a` flips it.
- Python stops early (short-circuits): in `a and b`, if `a` is false, `b` is never run. In `a or b`, if `a` is true, `b` is never run.
- `or` returns the first truthy value, so `name or "guest"` gives a default when `name` is empty.

```python
age, member = 20, False
print(age >= 18 and member, age >= 18 or member, not member)
print(False and (1 / 0))   # the division never runs
name = ""
print(name or "guest")
```
