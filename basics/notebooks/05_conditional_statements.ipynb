## 1. if and else

- `if condition:` runs the indented block only when the condition is true. Indent with 4 spaces.
- `else:` runs when it's false. The condition can be any value: empty and zero values count as false.
- A one-line choice: `value = a if condition else b`.

```python
temperature = 31
if temperature > 30:
    advice = "stay in the shade"
else:
    advice = "enjoy the sun"
print(advice)
print("even" if 7 % 2 == 0 else "odd")
```

## 2. elif chains

- `elif` adds more conditions. Python tests them top to bottom and runs only the first one that is true.
- Put the most specific (or highest) test first: check `>= 90` before `>= 80`.
- End with `else` to catch everything left over.

```python
score = 83
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
print(grade)
```

## 4. Guard clauses and match

- Handle bad or special cases first and `return` early; the main logic then needs no extra indentation.
- `raise ValueError("message")` stops a function when its input makes no sense.
- `match value:` with `case` lines compares one value against several options; `case _:` catches anything else.

```python
def describe(n):
    if n < 0:
        raise ValueError("n must not be negative")
    if n == 0:
        return "none"
    return "some"

print(describe(0), describe(5))

command = "stop"
match command:
    case "go":
        print("moving")
    case "stop":
        print("halting")
    case _:
        print("unknown")
```
