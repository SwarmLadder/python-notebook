# 05 — Conditional Statements

Programs often choose among several actions. Use explicit conditions, correct boundaries, and readable branches to make decisions.

*Basics · 60–90 min*

## You will learn

- Use comparison and logical expressions correctly

- Understand truthy/falsy values

## Examples

### 1. Basic if/else

An `if` block runs when its condition is truthy; `else` handles the alternative. Indentation defines block membership. A statement outside the block always runs after the decision.

```python
temperature = 15
if temperature < 18:
    print("Wear a jacket")
else:
    print("Light clothing is fine")
print("Advice complete")
```

**Check your understanding:** What prints when temperature is 22?

### 2. Ordered elif branches

Check the strongest threshold first because the first matching branch wins. Validate the accepted range before assigning grades; always test values around each boundary.

```python
score = 85
if not 0 <= score <= 100:
    label = "Invalid"
elif score >= 90:
    label = "A"
elif score >= 80:
    label = "B"
elif score >= 70:
    label = "C"
else:
    label = "Review required"
print(label)
```

**Check your understanding:** What happens for 89, 90, 0, 100, and -1?

### 3. Compound and guarded conditions

Use `and`, `or`, and `not` for multi-part tests. Short-circuiting can protect expressions, such as division, that would otherwise fail.

```python
age = 19
has_id = True
print("Allowed" if age >= 18 and has_id else "Denied")
denominator = 0
if denominator != 0 and 10 / denominator > 2:
    print("Large quotient")
else:
    print("No safe quotient to compare")
```

**Check your understanding:** Why is the denominator check first?

### 4. Truthy and falsy

Zero, empty strings, empty collections, and `None` are falsy. Most other values are truthy; a nonempty string like `"False"` is still truthy. Use explicit comparisons when they improve clarity.

```python
nickname = ""
display_name = nickname if nickname else "Guest"
print(display_name)
cart = []
if not cart:
    print("Cart is empty")
```

**Check your understanding:** Why is `if not cart:` simpler than `if len(cart) == 0:`?
