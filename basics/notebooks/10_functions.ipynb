## Essential truth

A call binds arguments to parameters and creates local bindings. Return sends a value to the caller and ends the call. Reaching the end returns None.

## 1. def and return

- `def name(parameters):` defines a function; the indented body runs each time you call `name(arguments)`.
- `return value` sends a result back and ends the function. With no `return`, the function gives `None`.
- Keep printing separate: return the value, and let the caller decide whether to print it.

```python
def area(width, height):
    return width * height

print(area(3, 4))

def shout(text):
    print(text.upper())

print(shout("hi"))   # prints HI, then None
```

## 2. Default and keyword arguments

- A default makes a parameter optional: `def price_with_tax(price, rate=0.2):`.
- Pass arguments by name in any order: `price_with_tax(rate=0.1, price=50)`.
- A `*` in the parameter list makes everything after it keyword-only: `def discount(price, *, percent):` must be called as `discount(80, percent=25)`.

```python
def price_with_tax(price, rate=0.2):
    return round(price * (1 + rate), 2)

print(price_with_tax(50), price_with_tax(50, 0.1), price_with_tax(rate=0.05, price=10))
```

## 4. Functions that decide

- Return from inside `if` blocks as soon as you know the answer.
- Loop with an early `return False` to test every item, then `return True` after the loop.
- A prime has no divisors between 2 and its square root, so test `range(2, int(n ** 0.5) + 1)`.

```python
def all_positive(values):
    for v in values:
        if v <= 0:
            return False
    return True

print(all_positive([1, 2]), all_positive([1, -2]))
```

```python
def acceptance_cases(submission):
    assert submission([(3, 2), (5, 1)]) == 11, 'Add every line cost'
    assert submission([(3, 2), (5, 1)], fee=4) == 15, 'Apply the fee once'
    assert submission([], 4) == 0, 'An empty order costs zero'
    assert submission([(7, 0)], 4) == 0, 'Zero-quantity lines are not orders'
    assert submission([(0, 2)], 4) == 4, 'A free item still counts as ordered'
    assert submission([(0, 2), (10, 0), (3, 4)], 5) == 17, 'Ignore zero quantities when deciding whether to charge a fee'
    lines = [[3, 2], [5, 1], [0, 2]]
    original = [[3, 2], [5, 1], [0, 2]]
    assert submission(lines, fee=4) == 15, 'Process list-based line items too'
    assert lines == original, 'Do not modify the caller\'s order lines'

check.assess('invoice_total', acceptance_cases)
```
