# 10 — Functions

Functions turn a sequence of statements into a reusable, testable operation. Prefer clear parameters and return values rather than hidden side effects.

*Basics · 60–90 min*

## You will learn

- Define and call a function
- Distinguish print from return
- Use default and keyword arguments
- Document behavior and test edge cases

## Examples

### 1. Defining functions

A function is declared with `def`, named parameters, and an indented body. `return` provides a result to the caller and stops execution. A function without explicit return produces `None`.

```python
def square(number):
    """Return the number multiplied by itself."""
    return number * number

print(square(7))
answer = square(3)
print(answer)
```

**Check your understanding:** Why would printing inside square make it less reusable?

### 2. Argument styles and defaults

Call a function with positional arguments, keyword arguments, or both in valid order. Defaults are useful when the caller usually wants a common option. Aim for a small and unsurprising signature.

```python
def describe_product(name, price, currency="RON"):
    return f"{name} costs {price:.2f} {currency}"

print(describe_product("Pen", 4.25))
print(describe_product(price=20, name="Notebook", currency="EUR"))
```

**Check your understanding:** Which parameters are required in this signature?

### 3. Pure functions and assertions

Pure functions depend on their inputs and return results without modifying external state. They are easier to reason about and test. `assert` is useful for learning tests but is not input validation in production programs.

```python
def is_even(value: int) -> bool:
    return value % 2 == 0

assert is_even(0) is True
assert is_even(7) is False
assert is_even(-4) is True
print("All checks passed")
```

**Check your understanding:** Which boundary cases would you test for an age classifier?

### 4. Multiple returns and mutable defaults

A function can return multiple values as a tuple. Avoid mutable defaults such as `items=[]`, which would be shared across calls. Use `None` and create a fresh list.

```python
def summarize(values):
    return (len(values), sum(values) / len(values)) if values else (0, None)

print(summarize([2, 4, 6]))
def add_item(value, items=None):
    if items is None:
        items = []
    items.append(value)
    return items
print(add_item("a"), add_item("b"))
```

**Check your understanding:** Why does an empty sequence need special handling when computing its average?
