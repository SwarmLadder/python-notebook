## Examples

### 1. Using print effectively

`print` accepts multiple arguments and inserts a separator between them. `end` controls the suffix, normally a newline. The last expression in a notebook cell can display a result, but standalone scripts should print deliberately.

```python
print("Python", "Basics", sep=" — ")
print("Loading", end="...")
print("done")
print("Line one\nLine two")
```

**Check your understanding:** What happens when `end=" "` is used for several print calls?

### 2. Receiving input in notebooks

`input(prompt)` reads one line and returns a string. To preserve a run-all workflow, the example uses a simulated response. Uncomment the marked line for an interactive experiment.

```python
# name = input("Your name: ")
name = "Alex"  # simulated input; comment out when testing interactively
print(f"Hello, {name.strip()}!")
```

**Check your understanding:** What does input return when the person only presses Enter?

### 3. Parsing typed numbers

The string ``"2" + "3"` produces ``"23"`, not 5. Use numeric conversion for arithmetic and handle invalid values using exceptions once you've studied lesson 12.

```python
# age_text = input("Your age: ")
age_text = "24"
age = int(age_text)
print("Next year:", age + 1)
print(f"Price: {float('15.25'):.2f}")
```

**Check your understanding:** What happens if `age_text` contains `"unknown"`?

### 4. Readable reports

A good output message names values, uses appropriate units, and avoids unnecessary precision. A well-labeled result is often more important than terse code.

```python
distance_km = 5.75
minutes = 43
pace = minutes / distance_km
print(f"Distance: {distance_km:.2f} km")
print(f"Duration: {minutes} min")
print(f"Pace: {pace:.1f} min/km")
```

**Check your understanding:** How might you explain the meaning of pace to a beginner?
