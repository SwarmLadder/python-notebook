### 1. Importing modules

`import math` loads a namespace, while `from math import sqrt` imports one name. Explicit names are easier to understand than wildcard imports. Standard-library modules ship with Python.

```python
import math
from statistics import mean, median
print(math.sqrt(81))
print(f"pi ≈ {math.pi:.4f}")
print(mean([2, 4, 8]), median([2, 4, 8]))
```

**Check your understanding:** Why might `math.sqrt()` be clearer than an unqualified sqrt?

### 2. Randomness and reproducibility

`random.Random(seed)` produces reproducible pseudo-random sequences for tests and games. `randint` includes both endpoints. For passwords, security tokens, or secrets, use the `secrets` module instead.

```python
import random
rng = random.Random(42)
print(rng.randint(1, 6))
print(rng.choice(["apple", "pear", "mango"]))
cards = [1, 2, 3, 4]
rng.shuffle(cards)
print(cards)
```

**Check your understanding:** Why would a fixed seed be inappropriate for generating security credentials?

### 3. Dates and durations

`datetime.date` models calendar dates and `timedelta` represents durations. Using these objects handles month and year boundaries without manual arithmetic.

```python
from datetime import date, timedelta
starting_date = date(2026, 11, 20)
print(starting_date.isoformat())
print((starting_date + timedelta(days=30)).isoformat())
```

**Check your understanding:** What happens when adding days across December 31?

### 4. Packages and environments

A module contains Python code; a package organizes modules. Third-party packages are installed into a Python environment. Check `sys.executable` to confirm which environment the notebook kernel uses. Requirements are listed in the root `requirements.txt`.

```python
import sys
import pathlib
print("Python version:", sys.version.split()[0])
print("Interpreter:", sys.executable)
print("Module:", pathlib.__name__)
```

**Check your understanding:** Why can a package install succeed in a terminal but fail to import in Jupyter?
