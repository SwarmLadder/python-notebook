## 2. Random numbers you can repeat

- `random.randint(1, 6)` gives a whole number from 1 to 6, both included; `random.choice(items)` picks one item.
- `random.Random(42)` makes a generator with a seed: the same seed always gives the same sequence, which makes tests repeatable.
- `random.shuffle(items)` shuffles a list in place.

```python
import random
rng = random.Random(42)
print([rng.randint(1, 6) for _ in range(5)])
rng = random.Random(42)
print([rng.randint(1, 6) for _ in range(5)])   # the same again
```

## 3. Dates and durations

- `datetime.date(2024, 3, 1)` is a date; subtracting dates gives a `timedelta` with `.days`.
- Add time with `timedelta(days=7)`.
- `date.fromisoformat("2024-03-01")` reads a date; `d.isoformat()` writes it back; `d.strftime("%d %b %Y")` formats it.

```python
from datetime import date, timedelta
start = date(2024, 3, 1)
end = date.fromisoformat("2024-03-15")
print((end - start).days, start + timedelta(days=7), start.strftime("%d %b %Y"))
```
