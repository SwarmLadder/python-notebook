## 3. Assertions

````python
def acceptance_cases(submission):
    counts = [3, 4, 2, 1]
    original = counts[:]
    assert submission(counts, 3) == 2, 'Return the first bin strictly below target'
    assert counts == original, 'Do not change the inventory input'
    assert submission([3], 3) is None, 'Matching the target is not a shortfall'
    assert submission([3, 3, 3], 3) is None, 'Never treat equality as an alert'
    assert submission([], 3) is None, 'An empty inventory has no shortfall'
    assert submission([4, 1, 0], 2) == 1, 'Return the first, not the last, shortfall'
    assert submission([-1, 5], 0) == 0, 'Handle readings below zero'

check.assess('first_shortfall', acceptance_cases)
````
