## Examples

### 1. Kinds of mistakes

Syntax errors prevent parsing, runtime errors interrupt execution, and logic errors return a wrong result with no exception. Begin with a small reproducible example and isolate the earliest wrong value.

```python
# Invalid syntax example (commented): if True print("missing colon")
# Runtime error example (commented): 10 / 0
width, height = 4, 6
wrong_area = width + height
correct_area = width * height
print(wrong_area, correct_area)
```

**Check your understanding:** Which kind of bug is easiest to miss without a test?

### 2. Reading error messages

The last line of a traceback identifies the exception class and message. Earlier frames show the call path. Catching a known exception in this educational cell keeps 'run all' safe.

```python
def average(values):
    return sum(values) / len(values)
for sample in [[1, 2, 3], []]:
    try:
        print(average(sample))
    except ZeroDivisionError as exc:
        print(type(exc).__name__, exc)
```

**Check your understanding:** How should average behave on an empty list?

### 3. Tracing state

Add temporary prints near suspected computations. Record state before and after an iteration, then compare your observations against a prediction. Remove temporary tracing after the bug is fixed.

```python
numbers = [3, 4, 5]
total = 0
for number in numbers:
    print("before:", total, "adding:", number)
    total += number
    print("after:", total)
print("result:", total)
```

**Check your understanding:** What should the accumulator equal after processing the first two numbers?

### 4. Assertions and notebooks

Assertions document expected outputs and can detect regressions. They are not a substitute for checking untrusted user inputs. Restart-and-run-all reveals stale variables left over from earlier exploratory execution.

```python
def double(n):
    return n * 2
assert double(0) == 0
assert double(-2) == -4
assert double(3) == 6
print("Tests pass")
```

**Check your understanding:** Which test would detect a function that returns abs(n*2) instead?
