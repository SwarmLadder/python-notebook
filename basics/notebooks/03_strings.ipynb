## 1. Indexing and slicing

- Positions start at 0. `s[0]` is the first character, `s[-1]` the last.
- `s[start:stop]` takes characters from `start` up to, but not including, `stop`. Leave either out to go to the edge: `s[:3]`, `s[3:]`.
- `s[::-1]` reverses. A slice past the end is fine (`s[:100]`); a single index past the end raises `IndexError`.

```python
language = "python"
print(language[0], language[-1], language[1:4], language[:2], language[2:], language[::-1])
print(language[:100])
```

## 3. Splitting, joining and searching

- `s.split()` splits on whitespace; `s.split(",")` splits on commas. The result is a list.
- `"-".join(words)` glues a list of strings with `-` between them.
- `s.find(x)` gives the position of `x` or `-1`; `s.count(x)` counts it; `s.startswith(x)` and `s.endswith(x)` test the ends.

```python
sentence = "learn python one rung at a time"
words = sentence.split()
print(words)
print("-".join(words))
print(sentence.find("python"), sentence.count("o"), sentence.endswith("time"))
```

## 4. Formatting with f-strings

- `f"{value}"` inserts a value; any expression works: `f"{price * 2}"`.
- After a colon comes the format: `:.2f` two decimals, `:,` thousands separators, `:.1%` a percentage, `:>8` right-align in 8 characters.
- Use `!r` to show the value's repr, with quotes: `f"{name!r}"`.

```python
price, ratio, population = 3.5, 0.256, 8_000_000
print(f"${price:.2f} | {ratio:.1%} | {population:,}")
print(f"[{'tea':>8}]  [{'tea':<8}]")
```

## 5. Checking characters

- `s.isdigit()` is true when every character is a digit; `s.isalpha()` when all are letters; `s.isalnum()` letters or digits. An empty string gives `False`.
- Compare text without caring about case with `a.casefold() == b.casefold()`.
- These checks help with input, but they aren't full validation: `"@" in email` does not make an email valid.

```python
print("2024".isdigit(), "12.5".isdigit(), "Ada".isalpha(), "R2D2".isalnum())
print("Straße".casefold() == "STRASSE".casefold())
```
