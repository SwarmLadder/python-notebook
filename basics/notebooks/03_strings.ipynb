## Examples

### 1. Indexing and slicing

Indexes start at zero; negative indexes count from the end. A slice `text[start:stop:step]` excludes the stop index. Slices are safe beyond sequence boundaries, while invalid individual indexes raise an error.

```python
language = "Python"
print(language[0], language[-1])
print(language[1:4], language[:3])
print(language[::-1])
print(len(language))
```

**Check your understanding:** Why does `language[:100]` work while `language[100]` fails?

### 2. Methods and immutability

String methods such as `strip`, `lower`, `replace`, `split`, `join`, and `find` return new values. They do not modify the original string. Normalize text before comparisons when whitespace or case can vary.

```python
raw = "  Python, Data, AI  "
clean = raw.strip().lower()
parts = [part.strip() for part in clean.split(",")]
print(parts)
print(" | ".join(parts))
print(raw)
```

**Check your understanding:** What is still stored in `raw` after `strip()`?

### 3. Formatting

F-strings interpolate expressions into text; format specifiers control decimal precision and alignment. Show currency, units, and labels to prevent ambiguity.

```python
item, price, count = "Notebook", 12.5, 3
print(f"{item}: {count} x {price:.2f} = {count * price:.2f}")
print(f"{'Item':<12}{'Price':>8}")
print(f"{item:<12}{price:>8.2f}")
```

**Check your understanding:** What format specifier would show a percentage with one decimal place?

### 4. Validation and Unicode

Methods including `isdigit`, `isalpha`, and `isalnum` answer specific questions about characters. For multilingual case-insensitive comparisons, `casefold()` is often more appropriate than `lower()`. A simplistic string check is not a complete email validator.

```python
user_text = "  Straße  "
print(user_text.strip().casefold())
print("123".isdigit())
print("test123".isalnum())
print("hello.py".endswith(".py"))
```

**Check your understanding:** Why might checking for an `@` character alone be insufficient to validate an email address?
