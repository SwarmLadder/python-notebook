### 1. Safe average

Treat empty input as a distinct valid case.

```python
def safe_average(values):
    if not values:
        return None
    return sum(values) / len(values)
assert safe_average([]) is None
assert safe_average([2, 4, 6]) == 4
```

**Check your understanding:** How should nonnumeric values be handled?

### 2. Validation with exceptions

A function can raise a clear exception, and its caller decides how to react.

```python
def discount(price, rate=0.1):
    if price < 0 or not 0 <= rate <= 1:
        raise ValueError("invalid price or rate")
    return price * (1 - rate)
assert discount(100) == 90
try:
    discount(-5)
except ValueError as error:
    print(error)
```

**Check your understanding:** Why validate before computing?

### 3. Temporary text file

Use pathlib and temporary directories for experiments that do not modify repository files.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
with TemporaryDirectory() as directory:
    path = Path(directory) / "notes.txt"
    path.write_text("alpha\nbeta\n", encoding="utf-8")
    assert path.read_text(encoding="utf-8").splitlines() == ["alpha", "beta"]
```

**Check your understanding:** What other cleanup is automatic?

### 4. CSV round trip

DictWriter and DictReader use consistent column names. Read numeric columns back as strings and then convert them.

```python
import csv
from pathlib import Path
from tempfile import TemporaryDirectory
with TemporaryDirectory() as directory:
    path = Path(directory) / "scores.csv"
    with path.open("w", newline="", encoding="utf-8") as stream:
        writer = csv.DictWriter(stream, fieldnames=["name", "grade"])
        writer.writeheader()
        writer.writerow({"name": "Ari", "grade": 91})
    with path.open(newline="", encoding="utf-8") as stream:
        rows = list(csv.DictReader(stream))
    assert int(rows[0]["grade"]) == 91
```

**Check your understanding:** What if a grade cell contains `N/A`?
