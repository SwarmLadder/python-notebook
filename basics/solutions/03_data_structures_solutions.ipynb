## Concepts and guided examples

Each concept starts with an explanation and a runnable example. The examples use Python's standard library so they work in a normal Jupyter Python 3 environment.

### 1. Second-largest distinct value

Deduplicate, sort, then check whether two distinct values exist.

```python
def second_largest(values):
    unique = sorted(set(values))
    return unique[-2] if len(unique) >= 2 else None
assert second_largest([3, 3, 7, 5]) == 5
assert second_largest([1, 1]) is None
```

**Check your understanding:** What does it mean for duplicates to be ignored?

### 2. Frequency table

The dictionary stores a running count for each normalized word.

```python
def word_counts(text):
    result = {}
    for word in text.casefold().split():
        result[word] = result.get(word, 0) + 1
    return result
print(word_counts("To be or not to be"))
```

**Check your understanding:** How should punctuation be handled?

### 3. Set overlap

Intersection and difference explicitly encode group comparisons.

```python
students_python = {"Ari", "Bea", "Cam"}
students_sql = {"Cam", "Dina"}
print(sorted(students_python & students_sql))
print(sorted(students_python - students_sql))
```

**Check your understanding:** What does union produce?

### 4. Nested dictionary averages

Iterate through inner records and guard empty score lists.

```python
records = {"S01": {"scores": [80, 90]}, "S02": {"scores": []}}
results = {}
for student_id, record in records.items():
    scores = record["scores"]
    results[student_id] = sum(scores) / len(scores) if scores else None
print(results)
```

**Check your understanding:** Could returning None be preferable to zero for no scores?
