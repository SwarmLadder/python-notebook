## Concepts and guided examples

Each concept starts with an explanation and a runnable example. The examples use Python's standard library so they work in a normal Jupyter Python 3 environment.

### 1. Price calculation

Convert string values before multiplication; format only for display.

```python
price_text = "7.25"
quantity = 7
result = float(price_text) * quantity
print(f"{result:.2f}")
assert result == 50.75
```

**Check your understanding:** What changes for fractional quantities?

### 2. Time conversion

Floor division finds whole hours, and modulo returns the remaining minutes.

```python
minutes = 187
hours = minutes // 60
remainder = minutes % 60
print(f"{hours} hours {remainder} minutes")
assert (hours, remainder) == (3, 7)
```

**Check your understanding:** Should negative minutes be accepted?

### 3. Text normalization

Strip unwanted padding and use casefold for broader case-insensitive text handling.

```python
raw = "  PYTHON, DATA, AI  "
topics = [part.strip().casefold() for part in raw.split(",")]
print(topics)
assert topics == ["python", "data", "ai"]
```

**Check your understanding:** What if the input contains empty fields?

### 4. Simple eligibility

Use chained comparisons to express a valid range clearly.

```python
def valid_score(score):
    return 0 <= score <= 100
assert valid_score(0) and valid_score(100)
assert not valid_score(-1) and not valid_score(101)
```

**Check your understanding:** What if a string is passed instead?
