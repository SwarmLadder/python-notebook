## Concepts and guided examples

Each concept starts with an explanation and a runnable example. The examples use Python's standard library so they work in a normal Jupyter Python 3 environment.

### 1. Classification

Order branch tests from most restrictive to more general.

```python
def grade(score):
    if not 0 <= score <= 100:
        return "invalid"
    if score >= 90:
        return "A"
    if score >= 80:
        return "B"
    if score >= 70:
        return "C"
    return "D"
assert grade(90) == "A"
assert grade(-1) == "invalid"
```

**Check your understanding:** Check the boundary on each branch.

### 2. Factorial

A product accumulator starts at 1. By convention 0! is 1.

```python
def factorial(n):
    if n < 0:
        raise ValueError("n must be nonnegative")
    result = 1
    for value in range(2, n + 1):
        result *= value
    return result
assert factorial(0) == 1
assert factorial(5) == 120
```

**Check your understanding:** Why does multiplying from 2 work?

### 3. Counting signs

Use mutually exclusive branches so each value is counted once.

```python
def count_signs(values):
    counts = {"positive": 0, "negative": 0, "zero": 0}
    for value in values:
        if value > 0:
            counts["positive"] += 1
        elif value < 0:
            counts["negative"] += 1
        else:
            counts["zero"] += 1
    return counts
print(count_signs([-1, 0, 2, 4]))
```

**Check your understanding:** Check how an empty list behaves.

### 4. FizzBuzz

Check the combined condition first so multiples of both 3 and 5 are handled properly.

```python
def fizzbuzz(n):
    if n % 15 == 0:
        return "FizzBuzz"
    if n % 3 == 0:
        return "Fizz"
    if n % 5 == 0:
        return "Buzz"
    return str(n)
print([fizzbuzz(n) for n in range(1, 21)])
```

**Check your understanding:** What happens if the 15 case is checked last?
