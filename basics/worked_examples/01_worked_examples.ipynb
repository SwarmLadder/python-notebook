## 1. Words longer than n letters

Count how many words in a sentence have more than `n` letters, ignoring punctuation.

**Plan**

1. Split the sentence into words.
2. Strip punctuation from each word.
3. Count the words whose length is greater than `n`.

```python
import string

def count_long_words(sentence, n):
    count = 0
    for word in sentence.split():
        word = word.strip(string.punctuation)
        if len(word) > n:
            count += 1
    return count

print(count_long_words("Python is fun, isn't it? Absolutely!", 3))
assert count_long_words("Python is fun, isn't it? Absolutely!", 3) == 3
assert count_long_words("", 3) == 0
```

**Why it works:** `strip(string.punctuation)` removes punctuation only at the ends, so `isn't` keeps its apostrophe and has 5 characters.

**Common mistake:** Counting `fun,` as 4 characters because the comma was not removed.

**Your turn:** Return the long words themselves, in order, instead of the count.

## 2. First time the balance goes negative

Given a starting balance and a list of transactions (+ deposits, − withdrawals), return the position (1-based) of the first transaction that makes the balance negative, or `None`.

**Plan**

Keep a running balance. After each transaction, check whether it dropped below zero, and stop as soon as it did.

```python
def first_overdraft(balance, transactions):
    for position, amount in enumerate(transactions, start=1):
        balance += amount
        if balance < 0:
            return position
    return None

print(first_overdraft(50, [20, -60, -20, 100]))
assert first_overdraft(50, [20, -60, -20, 100]) == 3
assert first_overdraft(10, [5, -15]) is None   # exactly 0 is not negative
```

**Why it works:** `enumerate(..., start=1)` gives human-friendly positions, and `return` inside the loop stops at the first overdraft.

**Common mistake:** Using `<= 0`, which treats a balance of exactly zero as an overdraft.

**Your turn:** Also return the balance at that moment, as a tuple `(position, balance)`.

## 3. Group names by first letter

Turn `['ana', 'Ben', 'bob', 'Cy']` into `{'A': ['ana'], 'B': ['Ben', 'bob'], 'C': ['Cy']}`.

**Plan**

Loop over the names. The key is the first letter in uppercase. Create an empty list the first time you see a key, then append.

```python
def group_by_initial(names):
    groups = {}
    for name in names:
        key = name[0].upper()
        if key not in groups:
            groups[key] = []
        groups[key].append(name)
    return groups

print(group_by_initial(["ana", "Ben", "bob", "Cy"]))
assert group_by_initial(["ana", "Ben", "bob"]) == {"A": ["ana"], "B": ["Ben", "bob"]}
assert group_by_initial([]) == {}
```

**Why it works:** A dict maps each letter to a list, so one pass over the names is enough.

**Common mistake:** Writing `groups[key] = [name]` every time, which replaces the list instead of adding to it.

**Your turn:** Rewrite it with `groups.setdefault(key, []).append(name)`.

## 4. Average per subject from text lines

Each line looks like `'math: 80'`. Compute the average score per subject, rounded to 1 decimal.

**Plan**

Split each line on `':'`, strip the spaces, convert the score with `int()`. Keep a total and a count per subject, then divide.

```python
def subject_averages(lines):
    totals, counts = {}, {}
    for line in lines:
        subject, score = line.split(":")
        subject = subject.strip()
        totals[subject] = totals.get(subject, 0) + int(score)
        counts[subject] = counts.get(subject, 0) + 1
    return {s: round(totals[s] / counts[s], 1) for s in totals}

data = ["math: 80", "art: 95", "math: 91", "art: 70", "math: 77"]
print(subject_averages(data))
assert subject_averages(data) == {"math": 82.7, "art": 82.5}
```

**Why it works:** Keeping totals and counts separately means each line is processed once. `int(' 80')` accepts surrounding spaces.

**Common mistake:** Averaging the running averages, which gives the wrong answer when subjects have different numbers of scores.

**Your turn:** Skip lines that are malformed (no `:` or not a number) instead of crashing.

## 5. Explain why a username is invalid

Return a list of every rule a username breaks: 3–12 characters, letters/digits/underscore only, must not start with a digit. An empty list means valid.

**Plan**

Check each rule independently and append a message for each failure. Don't stop at the first problem: users want all of them at once.

```python
def username_problems(name):
    problems = []
    if not 3 <= len(name) <= 12:
        problems.append("must be 3-12 characters")
    if not all(ch.isalnum() or ch == "_" for ch in name):
        problems.append("only letters, digits and _ allowed")
    if name[:1].isdigit():
        problems.append("must not start with a digit")
    return problems

for candidate in ["ada_99", "9lives", "x", "hi there!"]:
    print(repr(candidate), username_problems(candidate))
assert username_problems("ada_99") == []
assert username_problems("") == ["must be 3-12 characters"]
```

**Why it works:** Separate `if` statements (not `elif`) let every rule be reported. `name[:1]` is safe even for an empty string.

**Common mistake:** Using `name[0]`, which crashes on an empty username.

**Your turn:** Add a rule: the name must not be in a list of reserved words like `admin`.

## 6. Longest warming streak

Given daily temperatures, find the length of the longest run of days where each day is warmer than the day before.

**Plan**

Walk through consecutive pairs. Extend the current streak when the temperature rises, otherwise reset it. Remember the best streak seen.

```python
def longest_warming_streak(temps):
    if not temps:
        return 0
    best = current = 1
    for previous, today in zip(temps, temps[1:]):
        current = current + 1 if today > previous else 1
        best = max(best, current)
    return best

print(longest_warming_streak([12, 14, 15, 11, 13, 16, 18, 10]))
assert longest_warming_streak([12, 14, 15, 11, 13, 16, 18, 10]) == 4
assert longest_warming_streak([5, 5, 5]) == 1
assert longest_warming_streak([]) == 0
```

**Why it works:** `zip(temps, temps[1:])` pairs each day with the next, which avoids index arithmetic.

**Common mistake:** Forgetting to update `best` inside the loop, so a streak that ends at the last day is missed.

**Your turn:** Return the start and end positions of the streak too.
