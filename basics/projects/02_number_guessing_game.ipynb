## Examples

### 1. Game rules and validation

A target is chosen from 1 to 100. Each guess counts as one attempt if it is a valid in-range integer. The computer replies 'too low', 'too high', or 'correct'. A small function makes the comparison independently testable.

```python
def evaluate_guess(target, guess):
    if guess < target:
        return "too low"
    if guess > target:
        return "too high"
    return "correct"
assert evaluate_guess(50, 10) == "too low"
assert evaluate_guess(50, 90) == "too high"
assert evaluate_guess(50, 50) == "correct"
```

**Check your understanding:** Why validate guesses to be within the advertised range?

### 2. Run a simulated game

A fixed random seed makes an instructional demonstration repeatable. The following function processes an iterable of guesses and returns a transcript plus success status. It does not require user input.

```python
import random

def play_simulation(target, guesses):
    history = []
    for attempt, guess in enumerate(guesses, 1):
        if not isinstance(guess, int) or not 1 <= guess <= 100:
            history.append((attempt, guess, "invalid guess"))
            continue
        result = evaluate_guess(target, guess)
        history.append((attempt, guess, result))
        if result == "correct":
            break
    return history

rng = random.Random(42)
secret = rng.randint(1, 100)
print("Secret for demo:", secret)
for entry in play_simulation(secret, [25, 75, 50, secret]):
    print(entry)
```

### 3. Optional interactive version

Interactive notebook cells pause for answers. Keep the following loop as a reusable function and only call it manually when practicing. Catch malformed integers and allow the player to quit.

```python
def play_interactive(target):
    attempts = 0
    while True:
        response = input("Guess 1–100 (or q to quit): ").strip()
        if response.lower() == "q":
            print("Game ended")
            return False
        try:
            guess = int(response)
        except ValueError:
            print("Enter a whole number")
            continue
        if not 1 <= guess <= 100:
            print("Choose a number between 1 and 100")
            continue
        attempts += 1
        feedback = evaluate_guess(target, guess)
        print(feedback)
        if feedback == "correct":
            print(f"Solved in {attempts} attempts")
            return True

# Uncomment to play: play_interactive(random.randint(1, 100))
```
