## 01. Reusable iterable

A generator-based __iter__ returns a fresh generator each time.

**Acceptance criteria:** Two passes give the same values.

**Hint:** Do not return self unless the object owns iterator state.

```python
class EvenRange:
    def __init__(self, limit):
        self.limit = limit
    def __iter__(self):
        for n in range(0, self.limit + 1, 2):
            yield n
sequence = EvenRange(6)
assert list(sequence) == [0, 2, 4, 6]
assert list(sequence) == [0, 2, 4, 6]
```

## 02. Lazy Fibonacci

Yield pairs of values while updating local generator state.

**Acceptance criteria:** Handles n=0 correctly.

**Hint:** No entire sequence needs to be stored.

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
assert list(fibonacci(0)) == []
assert list(fibonacci(6)) == [0, 1, 1, 2, 3, 5]
```

## 03. Two independent counters

Use nonlocal to rebind captured state rather than a module-global counter.

**Acceptance criteria:** State isolated by factory call.

**Hint:** Each invocation creates new closure state.

```python
def make_counter(start=0):
    total = start
    def increment():
        nonlocal total
        total += 1
        return total
    return increment
a, b = make_counter(), make_counter(5)
assert (a(), a(), b()) == (1, 2, 6)
```

## 04. A metadata-preserving decorator

Preserve return values and attributes with wraps.

**Acceptance criteria:** Metadata retained and call count correct.

**Hint:** Wrapper should forward *args and **kwargs.

```python
from functools import wraps
def count_calls(func):
    state = {"count": 0}
    @wraps(func)
    def wrapper(*args, **kwargs):
        state["count"] += 1
        return func(*args, **kwargs)
    wrapper.call_state = state
    return wrapper
@count_calls
def add(a, b):
    """Add two numbers."""
    return a+b
assert add(2, 4) == 6
assert add.__name__ == "add"
assert add.call_state["count"] == 1
```

## 05. Cleanup context manager

Use try/finally so exit logic runs even if body raises.

**Acceptance criteria:** Exception propagates while cleanup executes.

**Hint:** finally is the critical piece.

```python
from contextlib import contextmanager
events = []
@contextmanager
def tracked():
    events.append("enter")
    try:
        yield
    finally:
        events.append("exit")
try:
    with tracked():
        raise ValueError("example")
except ValueError:
    pass
assert events == ["enter", "exit"]
```

## 06. Grouped totals

Sort by category before applying itertools.groupby.

**Acceptance criteria:** Unsorted input aggregated correctly.

**Hint:** groupby only combines adjacent equal keys.

```python
from itertools import groupby
rows = [("B", 2), ("A", 4), ("B", 3)]
output = {}
for category, group in groupby(sorted(rows), key=lambda row: row[0]):
    output[category] = sum(value for _, value in group)
assert output == {"A": 4, "B": 5}
print(output)
```
