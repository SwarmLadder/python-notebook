## 2. Groups

- Parentheses capture: `(\d{4})-(\d{2})` gives `m.group(1)`, `m.group(2)`.
- Name groups with `(?P<year>\d{4})`, then read `m["year"]` or `m.groupdict()`.
- Captured groups are strings; convert with `int()` if you need numbers.

```python
import re
m = re.search(r"(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})", "due 2024-03-15!")
print(m.group(0), m["year"], m.groupdict())
```

## 3. findall and sub

- `re.findall(pattern, text)` returns every match as a list of strings.
- `re.sub(pattern, replacement, text)` replaces every match.
- `[^a-z0-9]+` means one or more characters that are *not* lowercase letters or digits. Handy for slugs.

```python
import re
print(re.findall(r"#(\w+)", "#Py is fun #AI"))
print(re.sub(r"\d{3}-\d{4}", "[PHONE]", "call 555-1234 or 555-9876"))
print(re.sub(r"[^a-z0-9]+", "-", "hello, world!".lower()).strip("-"))
```
