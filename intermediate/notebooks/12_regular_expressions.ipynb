### 1. Searching and matching

re.search finds a matching substring, while re.fullmatch requires the entire input to match. Raw string literals beginning with r avoid accidental interpretation of backslashes. A match object's group and span methods provide the matched text and location.

```python
import re

text = "Order ID: A-1042"
match = re.search(r"[A-Z]-\d{4}", text)
print(match.group(0) if match else None)
print(bool(re.fullmatch(r"\d{4}", "2048")))
print(bool(re.fullmatch(r"\d{4}", "2048x")))
```

**Think about it:** Why is fullmatch more appropriate for a strict four-digit field?

### 2. Capture groups and named groups

Parentheses capture portions of a pattern, and named groups create self-documenting extraction. Character classes such as [A-Z] and quantifiers such as + and {2,4} express constraints, but syntax alone rarely captures all business rules.

```python
import re

record = "SKU=AB-204; QTY=12"
pattern = re.compile(r"SKU=(?P<sku>[A-Z]{2}-\d{3});\s*QTY=(?P<qty>\d+)")
match = pattern.fullmatch(record)
if match:
    print(match.groupdict())
    print(int(match.group("qty")))
```

**Think about it:** What would happen if the record had a negative quantity?

### 3. Extraction and substitution

findall extracts every match, while re.sub replaces matches using a pattern. These are useful for controlled normalization and pattern-based anonymization, but verify you are not unintentionally removing valuable text.

```python
import re

message = "Call 555-0100 or 555-0200"
numbers = re.findall(r"\b\d{3}-\d{4}\b", message)
redacted = re.sub(r"\b\d{3}-\d{4}\b", "[REDACTED]", message)
print(numbers)
print(redacted)
```

**Think about it:** How would this pattern behave with international phone formats?

### 4. Greedy vs non-greedy matching

Most quantifiers are greedy, consuming as much as possible while preserving a match. Adding ? can make them non-greedy. For general HTML, use an HTML parser rather than regex because nested markup needs structural parsing.

```python
import re

text = "<item>first</item><item>second</item>"
print(re.findall(r"<item>.*</item>", text))
print(re.findall(r"<item>.*?</item>", text))
print([word for word in re.split(r"[,;]\s*", "red, blue; green")])
```

**Think about it:** Why is a proper HTML parser more reliable for real webpages?
