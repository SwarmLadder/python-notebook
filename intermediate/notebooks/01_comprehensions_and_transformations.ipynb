## 3. Nested comprehensions and generators

- Two `for`s read left to right like nested loops: `[x for row in grid for x in row]` flattens.
- Transpose with an inner comprehension: `[[row[i] for row in grid] for i in range(len(grid[0]))]`.
- `sum(x * x for x in items)` uses a generator expression: no list is built in memory.

```python
grid = [[1, 2, 3], [4, 5, 6]]
print([x for row in grid for x in row])
print([[row[i] for row in grid] for i in range(3)])
print(sum(x * x for x in range(5)))
```
