# 01 — Comprehensions and Collection Transformations

Python comprehensions provide concise ways to construct lists, sets, and dictionaries. They are useful when the transformation is simple, but nested logic and side effects are usually clearer in ordinary loops.

*Intermediate · 75–120 min*

## You will learn

- Write list, set, and dictionary comprehensions
- Filter and transform values safely
- Understand nested comprehensions and evaluation order
- Recognize when a generator expression saves memory

## Examples

### 1. List comprehensions as a transformation

A list comprehension has the shape expression for item in iterable, followed by optional filtering conditions. It creates a list eagerly. Prefer a readable variable name and a simple expression; don't force complicated multi-step algorithms into a single line.

```python
prices = [12.5, 19.0, 30.5, 4.0]
discounted = [round(price * 0.9, 2) for price in prices]
expensive = [price for price in prices if price >= 15]
print(discounted)
print(expensive)
assert len(discounted) == len(prices)
```

**Reasoning prompt:** Why are the original prices unchanged?

### 2. Dictionary and set comprehensions

A dictionary comprehension creates key-value pairs and may overwrite earlier values if duplicate keys appear. A set comprehension creates unique elements, with no guaranteed iteration order. These make indexing and deduplicating data concise.

```python
names = ["Ada", "GRACE", "Ada", "Lin"]
normalized = {name.casefold() for name in names}
length_by_name = {name: len(name) for name in sorted(normalized)}
print(sorted(normalized))
print(length_by_name)
```

**Reasoning prompt:** What happens when two source records map to the same dictionary key?

### 3. Conditional output and nested loops

A conditional expression inside a comprehension chooses a value, while a trailing if decides whether an item is included. A nested comprehension visits its iterables in the same order as equivalent nested loops.

```python
numbers = list(range(-3, 4))
labels = ["positive" if n > 0 else "nonpositive" for n in numbers]
pairs = [(a, b) for a in (1, 2) for b in ("x", "y")]
print(list(zip(numbers, labels)))
print(pairs)
```

**Reasoning prompt:** Rewrite the pairs comprehension with two explicit for loops.

### 4. Generator expressions vs list comprehensions

A generator expression computes values when consumed instead of building a list immediately. This often saves memory for large data streams. Functions such as sum, any, and all consume iterables directly. Generators are usually single-use.

```python
values = range(1, 1001)
sum_squares = sum(n * n for n in values if n % 2 == 0)
print(sum_squares)
generator = (n * n for n in range(1, 4))
print(next(generator), list(generator))
print(list(generator))
```

**Reasoning prompt:** Why is the final list empty after the generator has been exhausted?

## Common mistakes

- A comprehension should express a transformation, not hide complex side effects.
- A generator is consumed as values are requested and cannot normally be rewound.
- Dictionary keys collide if the transformation produces duplicate keys.
