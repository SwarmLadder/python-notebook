## Examples

### 1. Growth rates and nested work

A loop over n items is typically O(n). Two nested loops over n items are typically O(n²). Membership tests in a list are O(n) in the worst case, while set membership is O(1) on average for typical hashable values. Constant factors still matter for small inputs.

```python
def count_pairs(values):
    count = 0
    for left in values:
        for right in values:
            count += 1
    return count

for n in (5, 10, 20):
    print(n, count_pairs(range(n)))
```

**Think about it:** What happens to the number of pair operations when n doubles?

### 2. Choose appropriate data structures

If you repeatedly check whether values belong to a large collection, converting to a set often improves lookup speed. Building the set itself costs time and memory, so consider the number of lookups and whether order or duplicates matter.

```python
allowed_list = list(range(10000))
allowed_set = set(allowed_list)
queries = [2, 50, 5000, 20000]
print([query in allowed_list for query in queries])
print([query in allowed_set for query in queries])
assert [query in allowed_list for query in queries] == [query in allowed_set for query in queries]
```

**Think about it:** Why not replace every list with a set?

### 3. Benchmarking with timeit

timeit repeats a small code fragment and measures elapsed time, reducing some measurement noise. Compare equivalent work and rerun measurements. Differences depend on your environment, Python version, and input distribution.

```python
import timeit

data = list(range(5000))
lookup_set = set(data)
list_duration = timeit.timeit(lambda: 4999 in data, number=100)
set_duration = timeit.timeit(lambda: 4999 in lookup_set, number=100)
print(f"List lookup: {list_duration:.6f}s")
print(f"Set lookup: {set_duration:.6f}s")
```

**Think about it:** How might lookup near the beginning of the list affect your conclusions?

### 4. Profiling and trade-offs

cProfile reports time spent in functions and how often they are called. Use it to identify bottlenecks before rewriting code. Faster implementations can reduce readability or increase memory use; weigh performance against maintainability.

```python
import cProfile
import io
import pstats

def workload():
    return sum(i * i for i in range(10000))

profiler = cProfile.Profile()
profiler.enable()
answer = workload()
profiler.disable()
stream = io.StringIO()
pstats.Stats(profiler, stream=stream).sort_stats("cumulative").print_stats(5)
print("Result:", answer)
print(stream.getvalue()[:800])
```

**Think about it:** Why would profiling one tiny run be less reliable than profiling realistic workloads?
