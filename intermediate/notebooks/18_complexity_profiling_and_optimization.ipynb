## 1. Big-O

- Big-O describes how work grows with input size `n`: O(1) constant, O(log n), O(n) one pass, O(n²) a loop inside a loop.
- Drop constants and small terms: two passes is still O(n).
- `x in list` is O(n); `x in set` or `in dict` is O(1) on average.

```python
def has_duplicates_slow(items):          # O(n^2)
    return any(items[i] == items[j] for i in range(len(items)) for j in range(i + 1, len(items)))

def has_duplicates_fast(items):          # O(n)
    return len(set(items)) != len(items)

print(has_duplicates_slow([1, 2, 3, 2]), has_duplicates_fast([1, 2, 3, 2]))
```

## 2. Better data structures

- Build a set or dict once, then look things up many times.
- `collections.Counter(items)` counts in one pass instead of calling `.count()` in a loop.
- `collections.deque` pops from the left in O(1); `list.pop(0)` is O(n).

```python
from collections import Counter, deque
words = "a b a c b a".split()
print(Counter(words).most_common(2))
q = deque([1, 2, 3])
q.popleft()
print(q)
```

## 3. Measure first

- Don't guess: measure. `timeit.timeit(stmt, number=...)` times a small snippet many times.
- `cProfile` shows which functions take the time; optimise the hotspot, not everything.
- Trade-offs: caching spends memory to save time; readability matters too.

```python
import timeit
setup = "data = list(range(5000)); s = set(data)"
print("list:", round(timeit.timeit("4999 in data", setup=setup, number=2000), 4))
print("set: ", round(timeit.timeit("4999 in s", setup=setup, number=2000), 4))
```
