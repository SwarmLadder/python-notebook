## 1. iter and next

- An iterable is anything a `for` loop accepts (lists, strings, dicts, files). `iter(x)` gets an iterator from it.
- `next(it)` returns the next item; when there are none left it raises `StopIteration`. `next(it, default)` returns `default` instead.
- An iterator is single-pass: once used up, it stays empty.

```python
it = iter([10, 20])
print(next(it), next(it), next(it, "done"))
squares = (x * x for x in range(3))
print(list(squares), list(squares))   # the second pass is empty
```

## 2. Writing an iterator class

- An iterator class has `__iter__` (returning `self`) and `__next__` (returning the next value or raising `StopIteration`).
- Keep the position in an attribute and update it in `__next__`.
- Because it returns `self`, such an object can be looped over only once.

```python
class Countdown:
    def __init__(self, start):
        self.current = start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current < 0:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value

print(list(Countdown(3)))
```

## 3. Reusable iterables and islice

- A reusable iterable's `__iter__` returns a fresh iterator each time, often by being a generator: `def __iter__(self): yield ...`.
- `itertools.islice(it, n)` takes the first `n` items, even from an endless iterator like `itertools.count()`.
- `zip(items, items[1:])` pairs neighbours; generalise it for sliding windows.

```python
from itertools import count, islice
class Squares:
    def __init__(self, n): self.n = n
    def __iter__(self):
        for i in range(self.n):
            yield i * i

s = Squares(4)
print(list(s), list(s))   # works twice
print(list(islice(count(10), 3)))
```
