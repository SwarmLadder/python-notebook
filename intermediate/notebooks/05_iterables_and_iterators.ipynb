# 05 — Iterables, Iterators, and Protocols

Iteration is one of Python's foundational protocols. Understanding iterables and iterators explains for loops, generators, comprehensions, and many standard library utilities.

## Examples

### 1. Iterable and iterator concepts

An iterable can produce an iterator. An iterator tracks progress and returns successive values via next until it raises StopIteration. Lists can create new iterators repeatedly; a generator is typically its own single-pass iterator.

```python
values = [10, 20, 30]
iterator = iter(values)
print(next(iterator))
print(next(iterator))
print(next(iterator, "finished"))
print(next(iterator, "finished"))
print(list(values))
```

**Reasoning prompt:** Why does iterating over values again work after exhausting iterator?

### 2. The iterator protocol in for loops

A for loop calls iter on its iterable and repeatedly calls next until StopIteration occurs. Most Python users should use for loops rather than next manually, but the model helps explain lazy iteration and debugging.

```python
class Countdown:
    def __init__(self, start):
        self.current = start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        answer = self.current
        self.current -= 1
        return answer

print(list(Countdown(4)))
```

**Reasoning prompt:** Why is Countdown itself an iterator rather than a reusable iterable?

### 3. Reusable iterable objects

A reusable iterable's __iter__ returns a new iterator each time. A generator function can implement this elegantly while avoiding manual state variables. Here each call to iter produces a fresh generator.

```python
class NumberRange:
    def __init__(self, start, stop):
        self.start, self.stop = start, stop
    def __iter__(self):
        for number in range(self.start, self.stop):
            yield number

numbers = NumberRange(2, 5)
print(list(numbers))
print(list(numbers))
```

**Reasoning prompt:** Why does this object work on both passes?

### 4. Lazy evaluation and exhaustion

Calling list or sum on an iterator consumes it. It is often a bug to inspect values by converting them to a list and then expect them to remain. Use itertools.tee with caution because it may buffer elements.

```python
iterator = iter([2, 4, 6])
print("sum:", sum(iterator))
print("remaining:", list(iterator))
values = [2, 4, 6]
print(sum(values), list(values))
```

## Common mistakes

- An iterator is not necessarily reusable or indexable.
- Converting an iterator to a list can use a large amount of memory.
- For most everyday code, generator functions are clearer than manually implementing __next__.
