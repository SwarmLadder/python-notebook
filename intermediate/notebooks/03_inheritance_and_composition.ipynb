## 1. Inheritance

- `class Circle(Shape):` makes `Circle` a subclass: it gets every method of `Shape` and can override any of them.
- `super().__init__(...)` runs the parent's `__init__` from the child's.
- `isinstance(obj, Shape)` is true for instances of `Shape` and of its subclasses.

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "..."

class Cat(Animal):
    def speak(self):
        return "meow"

tom = Cat("Tom")
print(tom.name, tom.speak(), isinstance(tom, Animal))
```

## 2. Polymorphism

- Different classes can offer the same method name; code that calls `shape.area()` works with any of them.
- A base method that subclasses must provide can `raise NotImplementedError`.
- Loop over a mixed list and let each object do its own thing: `sum(s.area() for s in shapes)`.

```python
import math
class Shape:
    def area(self):
        raise NotImplementedError
class Square(Shape):
    def __init__(self, side): self.side = side
    def area(self): return self.side ** 2
class Circle(Shape):
    def __init__(self, r): self.r = r
    def area(self): return math.pi * self.r ** 2

print(round(sum(s.area() for s in [Square(2), Circle(1)]), 2))
```

## 3. Composition

- Composition means an object holds other objects: a `Car` has an `Engine` rather than being one.
- Prefer composition when the relationship is "has a"; use inheritance for "is a".
- Delegate: the outer object calls methods on the inner one.

```python
class Engine:
    def start(self):
        return "vroom"

class Car:
    def __init__(self):
        self.engine = Engine()
    def start(self):
        return self.engine.start()

print(Car().start())
```
