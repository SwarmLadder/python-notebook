### 1. Inheritance and method overriding

A subclass inherits accessible methods from its base class and may override them. super() calls parent behavior cooperatively. Avoid deep inheritance hierarchies: they can make behavior difficult to locate and change.

```python
class Animal:
    def describe(self):
        return "an animal"

class Dog(Animal):
    def describe(self):
        return "a dog"

animals = [Animal(), Dog()]
print([animal.describe() for animal in animals])
```

**Reasoning prompt:** Which implementation runs when calling describe on a Dog object?

### 2. super and constructor validation

If the base constructor establishes invariants, the derived constructor should preserve them. Calling super().__init__ makes the relationship explicit. Favor readable arguments over delicate positional parameter arrangements.

```python
class Employee:
    def __init__(self, name):
        if not name:
            raise ValueError("name required")
        self.name = name

class Developer(Employee):
    def __init__(self, name, language):
        super().__init__(name)
        self.language = language
    def summary(self):
        return f"{self.name} uses {self.language}"

print(Developer("Ada", "Python").summary())
```

**Reasoning prompt:** What behavior is reused by super()?

### 3. Polymorphism without a shared ancestor

Python objects can be used polymorphically when they support the same operation. Duck typing emphasizes behavior rather than specific inheritance. Handle expected errors, and document the required interface.

```python
class EmailNotifier:
    def send(self, message):
        return f"email: {message}"

class ConsoleNotifier:
    def send(self, message):
        return f"console: {message}"

def notify_all(notifiers, message):
    return [notifier.send(message) for notifier in notifiers]

print(notify_all([EmailNotifier(), ConsoleNotifier()], "Hello"))
```

**Reasoning prompt:** Why is a common send method enough for this example?

### 4. Composition and dependency injection

A ReportService has a notifier rather than being a notifier. Passing dependencies from outside makes the service easier to test, substitute, and extend. This is a basic form of dependency injection.

```python
class ReportService:
    def __init__(self, notifier):
        self.notifier = notifier
    def publish(self, numbers):
        average = sum(numbers) / len(numbers) if numbers else None
        return self.notifier.send(f"average={average}")

reporter = ReportService(ConsoleNotifier())
print(reporter.publish([10, 20, 30]))
```

**Reasoning prompt:** What advantage does composition provide if you later change the delivery channel?

## Common errors and design choices

- Inheritance should model substitutability, not merely share a few lines of code.
- Overriding methods can violate guarantees provided by the base class.
- Composition is often easier to test and maintain than a deep hierarchy.
