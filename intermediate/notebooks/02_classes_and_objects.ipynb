## Examples

### 1. Classes, objects, and self

A class defines the behavior shared by instances. The __init__ method initializes a new object's attributes. self refers to the instance receiving a method call. Model a cohesive real-world concept rather than making classes for every tiny calculation.

```python
class BankAccount:
    def __init__(self, owner: str, balance: float = 0.0):
        if balance < 0:
            raise ValueError("opening balance cannot be negative")
        self.owner = owner
        self.balance = balance

    def deposit(self, amount: float) -> None:
        if amount <= 0:
            raise ValueError("deposit must be positive")
        self.balance += amount

account = BankAccount("Ada", 50)
account.deposit(25)
print(account.owner, account.balance)
```

**Reasoning prompt:** What happens to the account when deposit is called with a negative value?

### 2. Methods and encapsulated behavior

Move operations that must preserve object state into methods. Withdrawal is a good example: the method validates the amount and checks funds before changing balance. Python's public attributes remain accessible, so discipline and an explicit interface matter.

```python
class Wallet:
    def __init__(self, balance=0):
        self.balance = balance
    def spend(self, amount):
        if amount <= 0 or amount > self.balance:
            raise ValueError("invalid amount")
        self.balance -= amount
        return self.balance

wallet = Wallet(30)
print(wallet.spend(12))
try:
    wallet.spend(100)
except ValueError as exc:
    print(type(exc).__name__, str(exc))
```

**Reasoning prompt:** What additional type checks would a production financial app need?

### 3. Class attributes and instance attributes

A class attribute is looked up through the class and is often suitable for a shared constant. Assigning a value on an instance creates or replaces an instance attribute. Avoid mutable class-level defaults: a list defined on the class can accidentally be shared by all objects.

```python
class Counter:
    unit = "items"
    def __init__(self):
        self.value = 0
    def increment(self):
        self.value += 1

left, right = Counter(), Counter()
left.increment()
print(left.value, right.value, Counter.unit)
left.unit = "clicks"
print(left.unit, right.unit)
```

**Reasoning prompt:** Why is `items = []` at class scope dangerous for instance-specific items?

### 4. Object representation and assertions

Useful objects explain themselves. __repr__ helps debugging; use a deterministic representation showing key state and avoid leaking secrets. Assert expected behavior in normal and invalid scenarios.

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def distance_squared(self):
        return self.x ** 2 + self.y ** 2
    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

p = Point(3, 4)
print(p)
assert p.distance_squared() == 25
```

**Reasoning prompt:** What information should not appear in __repr__, even in a debugging log?

- self is an instance reference and must be named in instance method signatures.
