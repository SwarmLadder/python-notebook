### 1. Reproducible execution and environments

A notebook is executed by a Python kernel selected in Jupyter. Installing a package into the shell's Python does not guarantee it is installed in the running kernel. Inspect sys.executable and sys.version when troubleshooting. Prefer a repository-level virtual environment and a dependency file rather than global packages.

```python
import sys
from pathlib import Path
print("Python version:", sys.version.split()[0])
print("Kernel executable:", sys.executable)
print("Working directory:", Path.cwd())
print("This example does not need third-party packages.")
```

### 2. Functions first and explicit dependencies

Separate the core calculation from any notebook printing or input prompts. A function that receives everything it needs and returns a value can be reused in scripts, unit tests, CLI interfaces, or web services. Avoid hidden global state.

```python
def calculate_shipping(subtotal: float, *, free_over: float = 75.0) -> float:
    if subtotal < 0:
        raise ValueError("subtotal must not be negative")
    return 0.0 if subtotal >= free_over else 7.5

assert calculate_shipping(75) == 0
assert calculate_shipping(50) == 7.5
print(f"Shipping: {calculate_shipping(45):.2f}")
```

**Reasoning prompt:** Which test protects the inclusive boundary at 75?

### 3. Reusable modules and entry points

A module is usually a Python file imported by another file. Code guarded by if __name__ == '__main__' runs when the module is used as a script, not during a normal import. In notebooks, that guard may evaluate as true; here we demonstrate the value instead of running a CLI.

```python
def summarize(values):
    if not values:
        return None
    return {"count": len(values), "total": sum(values), "mean": sum(values) / len(values)}

print("__name__:", __name__)
print(summarize([10, 20, 30]))
```

**Reasoning prompt:** Why is performing network requests directly at import time considered risky?

### 4. Testing contracts and documentation

A function contract describes valid inputs, outputs, and errors. Type hints guide readers and static analyzers but do not enforce types at runtime. Assertions document example behavior; validate input explicitly where necessary.

```python
def percent_change(before: float, after: float) -> float:
    """Return fractional change; reject a zero baseline."""
    if before == 0:
        raise ValueError("before must be nonzero")
    return (after - before) / abs(before)

assert percent_change(100, 110) == 0.1
print(f"{percent_change(100, 115):.0%}")
```

**Reasoning prompt:** Would percent_change(0, 10) have a well-defined denominator?
