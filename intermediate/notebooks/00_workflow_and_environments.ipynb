## 2. Functions with explicit inputs

- A function should get everything it needs through its parameters and give results back with `return`, not through globals.
- Pure functions (same input → same output, no side effects) are the easiest to test and reuse.
- Do input/output (printing, files, network) at the edges and keep the logic in pure functions.

```python
def total(prices, tax_rate):
    return round(sum(prices) * (1 + tax_rate), 2)

print(total([10, 5], 0.2))
```

## 3. Modules and entry points

- A `.py` file is a module; `import tools` runs it once and gives you its names.
- Put script code under `if __name__ == "__main__":` so importing the module doesn't run it.
- `__name__` is `"__main__"` when the file is run directly, and the module's name when it's imported.

```python
def main():
    print("running as a script")

print(__name__)
if __name__ == "__main__":
    main()
```
